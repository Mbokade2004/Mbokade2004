```python
import random
import hashlib

# Function to perform modular exponentiation (for Diffie-Hellman)
def mod_exp(base, exp, mod):
    return pow(base, exp, mod)

# Function to simulate Diffie-Hellman Key Exchange for multiple parties
class ExtendedDiffieHellman:
    def __init__(self, p, g, num_parties):
        self.p = p    # Public prime number
        self.g = g    # Public generator
        self.num_parties = num_parties
        self.private_keys = []
        self.public_keys = []

    def generate_private_keys(self):
        """Generate private keys for all parties."""
        self.private_keys = [random.randint(2, self.p-1) for _ in range(self.num_parties)]

    def generate_public_keys(self):
        """Generate public keys for all parties based on private keys."""
        self.public_keys = [mod_exp(self.g, private_key, self.p) for private_key in self.private_keys]

    def calculate_shared_secret(self, party_index):
        """Calculate the shared secret for the given party."""
        shared_secret = 1
        for i in range(self.num_parties):
            if i != party_index:
                # Each party computes the secret using the public keys of others
                shared_secret = (shared_secret * mod_exp(self.public_keys[i], self.private_keys[party_index], self.p)) % self.p
        return shared_secret

    def print_details(self):
        """Print out the details of keys and shared secrets for each party."""
        print("Public Keys:")
        for i in range(self.num_parties):
            print(f"Party {i+1}: {self.public_keys[i]}")

        print("\nShared Secrets for each party:")
        for i in range(self.num_parties):
            shared_secret = self.calculate_shared_secret(i)
            print(f"Party {i+1}: {shared_secret}")


# Parameters for Diffie-Hellman
p = 23  # A small prime for simplicity (in real scenarios, p would be much larger)
g = 5   # Generator

# Number of parties involved in the key exchange
num_parties = 4  # You can change this number to simulate more parties

# Initialize Extended Diffie-Hellman key exchange
edhke = ExtendedDiffieHellman(p, g, num_parties)

# Step 1: Generate private keys for all parties
edhke.generate_private_keys()

# Step 2: Generate public keys based on private keys
edhke.generate_public_keys()

# Step 3: Print out public keys and shared secrets for each party
edhke.print_details()
```
