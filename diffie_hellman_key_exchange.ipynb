```python
import random

def diffie_hellman_key_exchange():
    # Prime number (p) and primitive root modulo p (g) - publicly agreed upon
    p = 23  # A small prime number
    g = 5   # A primitive root modulo 23

    print(f"Publicly shared values: p = {p}, g = {g}")

    # Alice chooses a private key (a) and calculates her public value (A)
    a = random.randint(1, p - 1)
    A = pow(g, a, p)
    print(f"Alice's private key (a): {a}")
    print(f"Alice's public value (A): {A}")

    # Bob chooses a private key (b) and calculates his public value (B)
    b = random.randint(1, p - 1)
    B = pow(g, b, p)
    print(f"Bob's private key (b): {b}")
    print(f"Bob's public value (B): {B}")

    # Exchange public values A and B over the public channel

    # Alice calculates the shared secret using Bob's public value
    shared_secret_alice = pow(B, a, p)
    print(f"Shared secret calculated by Alice: {shared_secret_alice}")

    # Bob calculates the shared secret using Alice's public value
    shared_secret_bob = pow(A, b, p)
    print(f"Shared secret calculated by Bob: {shared_secret_bob}")

    # Verify that both shared secrets are the same
    assert shared_secret_alice == shared_secret_bob, "Key exchange failed: Shared secrets do not match."

    print(f"Key exchange successful! Shared secret: {shared_secret_alice}")

# Simulate the Diffie-Hellman key exchange process
diffie_hellman_key_exchange()
```
